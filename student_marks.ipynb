```python
# Python program to calculate a student's average

student_name = "Arun"
marks = [85, 90, 78, 92, 88]

average = sum(marks) / len(marks)

print("Student:", student_name)
print("Average marks:", average)

if average >= 80:
    print("Result: Excellent")
else:
    print("Result: Keep practising")
```
